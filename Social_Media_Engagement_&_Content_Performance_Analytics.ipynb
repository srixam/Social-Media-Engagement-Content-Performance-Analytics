```python
# STEP 1: Download the original dataset

import pandas as pd
import requests
import hashlib
import os

# Original dataset URL
url = "https://raw.githubusercontent.com/abhishek07ma/Social_Media_Engagement_Dashboard/main/Social_Media_Engagement.csv"

# Download dataset
response = requests.get(url)
response.raise_for_status()

# Save original dataset without modifications
file_name = "Social_Media_Engagement_Original.csv"

with open(file_name, "wb") as file:
    file.write(response.content)

# Calculate file checksum
file_hash = hashlib.sha256(response.content).hexdigest()

print("Dataset downloaded successfully!")
print("File name:", file_name)
print("File size:", round(os.path.getsize(file_name) / 1024, 2), "KB")
print("SHA256:", file_hash)

# Load original dataset
df_original = pd.read_csv(file_name)

print("\nDataset loaded successfully!")
print("Number of rows:", df_original.shape[0])
print("Number of columns:", df_original.shape[1])

# Display first five records
display(df_original.head())
```

```python
# STEP 2: DATASET STRUCTURE UNDERSTANDING

import pandas as pd
import numpy as np

# 1. Display dataset dimensions
print("========== DATASET DIMENSIONS ==========")
print("Total Rows:", df_original.shape[0])
print("Total Columns:", df_original.shape[1])

# 2. Display column names
print("\n========== COLUMN NAMES ==========")
print(df_original.columns.tolist())

# 3. Display dataset information
print("\n========== DATASET INFORMATION ==========")
df_original.info()

# 4. Display first 10 records
print("\n========== FIRST 10 RECORDS ==========")
display(df_original.head(10))

# 5. Display last 5 records
print("\n========== LAST 5 RECORDS ==========")
display(df_original.tail())

# 6. Display statistical summary
print("\n========== STATISTICAL SUMMARY ==========")
display(df_original.describe(include='all').T)

# 7. Display data types
print("\n========== DATA TYPES ==========")
display(df_original.dtypes)

# 8. Count unique values in each column
print("\n========== UNIQUE VALUES ==========")
display(df_original.nunique(dropna=True))
```

```python
# STEP 3: PRELIMINARY DATA DICTIONARY

data_dictionary = pd.DataFrame({
    "Column Name": df_original.columns,
    "Data Type": df_original.dtypes.astype(str).values,
    "Non-Null Count": df_original.notnull().sum().values,
    "Missing Count": df_original.isnull().sum().values,
    "Unique Values": df_original.nunique().values
})

display(data_dictionary)

# Save data dictionary
data_dictionary.to_csv(
    "Preliminary_Data_Dictionary.csv",
    index=False
)

print("Data dictionary created successfully!")
```

```python
# STEP 4: MISSING VALUE ANALYSIS

import pandas as pd
import numpy as np

# Total missing values
print("========== MISSING VALUE ANALYSIS ==========")

total_missing = df_original.isnull().sum().sum()

print("Total Missing Values:", total_missing)

# Missing values in each column
missing_analysis = pd.DataFrame({
    "Column": df_original.columns,
    "Missing Values": df_original.isnull().sum().values,
    "Missing Percentage": (
        df_original.isnull().mean().values * 100
    ).round(2)
})

missing_analysis = missing_analysis.sort_values(
    by="Missing Percentage",
    ascending=False
)

display(missing_analysis)

# Visualize missing values
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))

sns.heatmap(
    df_original.isnull(),
    cbar=False,
    cmap="viridis",
    yticklabels=False
)

plt.title("Missing Values Heatmap")
plt.xlabel("Dataset Columns")
plt.ylabel("Dataset Records")
plt.show()
```

```python
# STEP 5: DUPLICATE RECORD ANALYSIS

# Count exact duplicate rows
total_duplicates = df_original.duplicated().sum()

print("========== DUPLICATE ANALYSIS ==========")

print("Total Records:", len(df_original))
print("Exact Duplicate Records:", total_duplicates)

print(
    "Duplicate Percentage:",
    round((total_duplicates / len(df_original)) * 100, 2),
    "%"
)

# Display duplicate records
duplicate_records = df_original[
    df_original.duplicated(keep=False)
]

print("\nDuplicate Records:")
display(duplicate_records.head(10))
```

```python
# STEP 6: CHECK UNIQUE CATEGORICAL VALUES

categorical_columns = df_original.select_dtypes(
    include=["object", "category"]
).columns

print("========== CATEGORICAL VALUE ANALYSIS ==========")

for column in categorical_columns:

    print("\nColumn:", column)

    print("Unique Values:")
    print(df_original[column].value_counts(dropna=False).head(30))
```

```python
# STEP 7: NUMERICAL OUTLIER ANALYSIS

numerical_columns = df_original.select_dtypes(
    include=np.number
).columns

outlier_results = []

for column in numerical_columns:

    Q1 = df_original[column].quantile(0.25)
    Q3 = df_original[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df_original[
        (df_original[column] < lower_limit) |
        (df_original[column] > upper_limit)
    ]

    outlier_results.append({
        "Column": column,
        "Lower Limit": lower_limit,
        "Upper Limit": upper_limit,
        "Outlier Count": len(outliers)
    })

outlier_report = pd.DataFrame(outlier_results)

display(outlier_report)
```

```python
# STEP 8: NEGATIVE VALUE CHECK

negative_report = {}

for column in numerical_columns:

    negative_count = (df_original[column] < 0).sum()

    negative_report[column] = negative_count

negative_df = pd.DataFrame(
    list(negative_report.items()),
    columns=["Column", "Negative Value Count"]
)

display(negative_df)
```

```python
# STEP 9: DATE COLUMN IDENTIFICATION

print("========== DATE COLUMN ANALYSIS ==========")

for column in df_original.columns:

    if any(word in column.lower() for word in
           ["date", "time", "timestamp", "created", "published"]):

        print("\nColumn:", column)
        print("Data Type:", df_original[column].dtype)
        print("Sample Values:")
        print(df_original[column].head(10).tolist())
```

```python
# STEP 10: DATE VALIDATION

date_columns = [
    column for column in df_original.columns
    if any(word in column.lower() for word in
           ["date", "time", "timestamp", "created", "published"])
]

date_validation = []

for column in date_columns:

    converted_dates = pd.to_datetime(
        df_original[column],
        errors="coerce"
    )

    invalid_dates = (
        df_original[column].notna() &
        converted_dates.isna()
    )

    date_validation.append({
        "Column": column,
        "Valid Dates": converted_dates.notna().sum(),
        "Invalid Dates": invalid_dates.sum(),
        "Earliest Date": converted_dates.min(),
        "Latest Date": converted_dates.max()
    })

date_validation_df = pd.DataFrame(date_validation)

display(date_validation_df)
```

```python
# STEP 11: LOGICAL CONSISTENCY CHECK

numeric_columns = df_original.select_dtypes(
    include=np.number
).columns

logical_report = []

for column in numeric_columns:

    logical_report.append({
        "Column": column,
        "Zero Values": (df_original[column] == 0).sum(),
        "Negative Values": (df_original[column] < 0).sum(),
        "Maximum Value": df_original[column].max(),
        "Minimum Value": df_original[column].min()
    })

logical_report_df = pd.DataFrame(logical_report)

display(logical_report_df)
```

```python
# STEP 12: COMPLETE DATA AUDIT REPORT

audit_report = pd.DataFrame({
    "Column": df_original.columns,
    "Data Type": df_original.dtypes.astype(str).values,
    "Total Records": len(df_original),
    "Non-Null Values": df_original.notnull().sum().values,
    "Missing Values": df_original.isnull().sum().values,
    "Missing Percentage": (
        df_original.isnull().mean().values * 100
    ).round(2),
    "Unique Values": df_original.nunique().values
})

print("========== COMPLETE DATA AUDIT REPORT ==========")

display(audit_report)

# Save report
audit_report.to_csv(
    "Complete_Data_Audit_Report.csv",
    index=False
)

print("Audit report saved successfully!")
```

```python
# STEP 14: CREATE WORKING COPY

import pandas as pd
import numpy as np

# Keep original dataset unchanged
df = df_original.copy(deep=True)

print("Original dataset rows:", len(df_original))
print("Working dataset rows:", len(df))

print("\nWorking copy created successfully!")
```

```python
# STEP 15: STANDARDISE COLUMN NAMES

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(r'[^a-zA-Z0-9]+', '_', regex=True)
    .str.strip('_')
)

print("Updated column names:")
print(df.columns.tolist())

print("\nColumn names standardised successfully!")
```

```python
# STEP 16: REMOVE EXACT DUPLICATES

before_rows = len(df)

df = df.drop_duplicates()

after_rows = len(df)

removed_duplicates = before_rows - after_rows

print("Rows before cleaning:", before_rows)
print("Duplicate rows removed:", removed_duplicates)
print("Rows after cleaning:", after_rows)
```

```python
# STEP 17: CLEAN TEXT COLUMNS

text_columns = df.select_dtypes(
    include=["object", "category"]
).columns

for column in text_columns:

    df[column] = df[column].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

    df[column] = df[column].replace("", np.nan)

print("Text columns cleaned successfully!")

display(df.head())
```

```python
# STEP 18: STANDARDISE CATEGORICAL VALUES

for column in text_columns:

    df[column] = df[column].apply(
        lambda x: x.title() if isinstance(x, str) else x
    )

print("Categorical formatting completed!")

for column in text_columns:
    print("\n", column)
    print(df[column].value_counts(dropna=False).head(15))
```

```python
# STEP 19: CONVERT DATE COLUMNS

date_columns = [
    col for col in df.columns
    if any(word in col for word in
           ["date", "time", "timestamp", "created", "published"])
]

print("Detected date columns:", date_columns)

for column in date_columns:

    df[column] = pd.to_datetime(
        df[column],
        errors="coerce"
    )

print("\nDate conversion completed!")

display(df[date_columns].head() if date_columns else
        pd.DataFrame({"Message": ["No date columns detected"]}))
```

```python
# STEP 20: CLEANED DATASET QUALITY CHECK

print("========== CLEANED DATASET SUMMARY ==========")

print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])

print("\nMissing values:")
display(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nData types:")
display(df.dtypes)

print("\nFirst five cleaned records:")
display(df.head())
```

```python
# STEP 21: SAVE CLEANED DATASET

cleaned_file = "Social_Media_Engagement_Cleaned.csv"

df.to_csv(cleaned_file, index=False)

print("Cleaned dataset saved successfully!")
print("File name:", cleaned_file)

from google.colab import files

files.download(cleaned_file)
```

```python
# STEP 22: MISSING VALUE ANALYSIS

missing_report = pd.DataFrame({
    "Column": df.columns,
    "Missing_Count": df.isnull().sum().values,
    "Missing_Percentage": (
        df.isnull().mean().values * 100
    ).round(2)
})

display(
    missing_report.sort_values(
        by="Missing_Percentage",
        ascending=False
    )
)
```

```python
# STEP 23: IDENTIFY COLUMN TYPES

numeric_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns.tolist()

datetime_columns = df.select_dtypes(
    include=["datetime"]
).columns.tolist()

print("Numerical Columns:")
print(numeric_columns)

print("\nCategorical Columns:")
print(categorical_columns)

print("\nDate Columns:")
print(datetime_columns)
```

```python
# STEP 24: MISSING VALUE TREATMENT

df_clean = df.copy()

# Numerical missing values
for column in numeric_columns:

    if df_clean[column].isnull().any():

        median_value = df_clean[column].median()

        if pd.notna(median_value):
            df_clean[column] = df_clean[column].fillna(
                median_value
            )

# Categorical missing values
for column in categorical_columns:

    if df_clean[column].isnull().any():

        mode_values = df_clean[column].mode()

        if not mode_values.empty:
            df_clean[column] = df_clean[column].fillna(
                mode_values[0]
            )

print("Missing value treatment completed.")

print("\nRemaining missing values:")
display(df_clean.isnull().sum())
```

```python
# STEP 25: SOCIAL MEDIA METRIC VALIDATION

engagement_keywords = [
    "like", "comment", "share",
    "view", "reach", "impression",
    "engagement"
]

engagement_columns = [
    column for column in df_clean.columns
    if any(word in column.lower()
           for word in engagement_keywords)
]

print("Detected engagement columns:")
print(engagement_columns)

for column in engagement_columns:

    if pd.api.types.is_numeric_dtype(df_clean[column]):

        print("\nColumn:", column)

        print("Minimum:", df_clean[column].min())
        print("Maximum:", df_clean[column].max())

        print(
            "Negative values:",
            (df_clean[column] < 0).sum()
        )
```

```python
# STEP 26: FINAL MISSING VALUE CHECK

remaining_missing = df_clean.isnull().sum().sum()

print("Total rows:", len(df_clean))
print("Total columns:", len(df_clean.columns))
print("Remaining missing values:", remaining_missing)

if remaining_missing == 0:
    print("No missing values remain.")
else:
    print("Some columns still contain missing values.")
    display(df_clean.isnull().sum())
```

```python
# STEP 27: SAVE PROVISIONALLY CLEANED DATASET

df_clean.to_csv(
    "Social_Media_Engagement_Provisional_Cleaned.csv",
    index=False
)

print("Provisional cleaned dataset saved successfully!")
```

```python
# STEP 28: INSPECT CLEANED DATASET

print("Dataset Shape:", df_clean.shape)

print("\nAvailable Columns:")
for col in df_clean.columns:
    print(col, ":", df_clean[col].dtype)

display(df_clean.head())
```

```python
# STEP 29: IDENTIFY ENGAGEMENT METRICS

columns = df_clean.columns.tolist()

possible_metrics = {
    "likes": ["likes", "like_count", "num_likes"],
    "comments": ["comments", "comment_count", "num_comments"],
    "shares": ["shares", "share_count", "num_shares"],
    "views": ["views", "view_count", "video_views"],
    "reach": ["reach", "total_reach"],
    "impressions": ["impressions", "impression_count"]
}

detected_metrics = {}

for metric, possible_names in possible_metrics.items():

    detected = [
        col for col in columns
        if col.lower() in possible_names
    ]

    detected_metrics[metric] = detected[0] if detected else None

print("Detected Metrics:")

for metric, col in detected_metrics.items():
    print(metric, ":", col)
```

```python
# STEP 30: CALCULATE TOTAL ENGAGEMENT

engagement_components = [
    detected_metrics["likes"],
    detected_metrics["comments"],
    detected_metrics["shares"]
]

engagement_components = [
    col for col in engagement_components
    if col is not None
]

if engagement_components:

    df_clean["total_engagement"] = (
        df_clean[engagement_components]
        .sum(axis=1, min_count=1)
    )

    print("Total engagement calculated successfully!")

    display(
        df_clean[
            engagement_components + ["total_engagement"]
        ].head(10)
    )

else:
    print("No suitable engagement columns detected.")
```

```python
# STEP 31: CALCULATE ENGAGEMENT RATE

denominator = None

if detected_metrics["reach"] is not None:
    denominator = detected_metrics["reach"]
elif detected_metrics["impressions"] is not None:
    denominator = detected_metrics["impressions"]

if denominator and "total_engagement" in df_clean.columns:

    valid_denominator = df_clean[denominator] > 0

    df_clean["engagement_rate"] = np.where(
        valid_denominator,
        (df_clean["total_engagement"] /
         df_clean[denominator]) * 100,
        np.nan
    )

    print("Engagement rate calculated using:", denominator)

    display(
        df_clean[
            ["total_engagement", "engagement_rate"]
        ].head(10)
    )

else:
    print("Engagement rate cannot be calculated.")
    print("Required engagement and denominator fields are unavailable.")
```

```python
# STEP 32: DATE FEATURE ENGINEERING

date_candidates = [
    col for col in df_clean.columns
    if any(word in col.lower()
           for word in ["date", "timestamp", "published", "created"])
]

print("Possible date columns:", date_candidates)

if date_candidates:

    date_col = date_candidates[0]

    df_clean[date_col] = pd.to_datetime(
        df_clean[date_col],
        errors="coerce"
    )

    df_clean["post_year"] = df_clean[date_col].dt.year
    df_clean["post_month"] = df_clean[date_col].dt.month
    df_clean["post_day"] = df_clean[date_col].dt.day

    df_clean["post_day_name"] = (
        df_clean[date_col].dt.day_name()
    )

    print("Date features created using:", date_col)

    display(df_clean.head())

else:
    print("No date column detected. Date features skipped.")
```

```python
# STEP 33: POST PERFORMANCE SEGMENTATION

if "total_engagement" in df_clean.columns:

    valid_values = df_clean["total_engagement"].dropna()

    if valid_values.nunique() >= 4:

        df_clean["performance_category"] = pd.qcut(
            df_clean["total_engagement"],
            q=4,
            labels=[
                "Low",
                "Medium",
                "High",
                "Very High"
            ],
            duplicates="drop"
        )

        print("Performance categories created!")

        display(
            df_clean["performance_category"]
            .value_counts()
        )

    else:
        print("Insufficient distinct values for quartile segmentation.")

else:
    print("Total engagement is not available.")
```

```python
# STEP 34: KPI SUMMARY

kpi_summary = {}

if "total_engagement" in df_clean.columns:

    kpi_summary["Average Engagement"] = (
        df_clean["total_engagement"].mean()
    )

    kpi_summary["Maximum Engagement"] = (
        df_clean["total_engagement"].max()
    )

    kpi_summary["Minimum Engagement"] = (
        df_clean["total_engagement"].min()
    )

if "engagement_rate" in df_clean.columns:

    kpi_summary["Average Engagement Rate (%)"] = (
        df_clean["engagement_rate"].mean()
    )

kpi_df = pd.DataFrame(
    list(kpi_summary.items()),
    columns=["KPI", "Value"]
)

display(kpi_df)
```

```python
# STEP 35: SAVE FEATURE-ENGINEERED DATASET

df_clean.to_csv(
    "Social_Media_Feature_Engineered.csv",
    index=False
)

kpi_df.to_csv(
    "Social_Media_KPI_Summary.csv",
    index=False
)

print("Feature-engineered dataset saved!")
print("KPI summary saved!")

print("Final dataset shape:", df_clean.shape)
```

```python
# STEP 48: SELECT TWO PLATFORMS

if platform_candidates and "total_engagement" in df_clean.columns:

    platform_counts = df_clean[platform_col].value_counts()

    if len(platform_counts) >= 2:

        selected_platforms = platform_counts.index[:2].tolist()

        group1 = df_clean.loc[
            df_clean[platform_col] == selected_platforms[0],
            "total_engagement"
        ].dropna()

        group2 = df_clean.loc[
            df_clean[platform_col] == selected_platforms[1],
            "total_engagement"
        ].dropna()

        print("Group 1:", selected_platforms[0])
        print("Group 2:", selected_platforms[1])

        print("Group 1 sample size:", len(group1))
        print("Group 2 sample size:", len(group2))

    else:
        print("Not enough platforms for comparison.")

else:
    print("Required columns unavailable.")
```

```python
# STEP 49: NORMALITY TEST

if len(group1) >= 3 and len(group2) >= 3:

    sample1 = group1.sample(
        min(len(group1), 5000),
        random_state=42
    )

    sample2 = group2.sample(
        min(len(group2), 5000),
        random_state=42
    )

    shapiro1 = stats.shapiro(sample1)
    shapiro2 = stats.shapiro(sample2)

    print("Group 1 Shapiro p-value:", shapiro1.pvalue)
    print("Group 2 Shapiro p-value:", shapiro2.pvalue)

    print("\nInterpretation:")
    if shapiro1.pvalue > 0.05 and shapiro2.pvalue > 0.05:
        print("No strong evidence against normality.")
    else:
        print("At least one group shows evidence of non-normality.")

else:
    print("Insufficient observations for normality testing.")
```

```python
# STEP 50: HYPOTHESIS TESTING

alpha = 0.05

if len(group1) >= 2 and len(group2) >= 2:

    if (
        len(group1) >= 3 and
        len(group2) >= 3 and
        shapiro1.pvalue > alpha and
        shapiro2.pvalue > alpha
    ):

        test_result = stats.ttest_ind(
            group1,
            group2,
            equal_var=False
        )

        test_name = "Welch Independent Samples T-Test"

    else:

        test_result = stats.mannwhitneyu(
            group1,
            group2,
            alternative="two-sided"
        )

        test_name = "Mann-Whitney U Test"

    print("Test Used:", test_name)
    print("Test Statistic:", test_result.statistic)
    print("P-value:", test_result.pvalue)

    if test_result.pvalue < alpha:
        print("Result: Reject the null hypothesis.")
        print("There is evidence of a difference between groups.")
    else:
        print("Result: Fail to reject the null hypothesis.")
        print("There is insufficient evidence of a difference.")

else:
    print("Insufficient data for hypothesis testing.")
```

```python
# STEP 51: CONFIDENCE INTERVALS

def calculate_ci(data):

    n = len(data)
    mean = data.mean()
    std = data.std(ddof=1)

    if n < 2:
        return np.nan, np.nan, np.nan

    standard_error = std / np.sqrt(n)

    margin = stats.t.ppf(
        0.975,
        df=n-1
    ) * standard_error

    return mean, mean-margin, mean+margin


ci1 = calculate_ci(group1)
ci2 = calculate_ci(group2)

confidence_report = pd.DataFrame({
    "Platform": selected_platforms,
    "Mean Engagement": [ci1[0], ci2[0]],
    "95% CI Lower": [ci1[1], ci2[1]],
    "95% CI Upper": [ci1[2], ci2[2]]
})

display(confidence_report)
```

```python
# STEP 52: PLATFORM ENGAGEMENT COMPARISON

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df_clean[
        df_clean[platform_col].isin(selected_platforms)
    ],
    x=platform_col,
    y="total_engagement"
)

plt.title("Engagement Distribution Across Selected Platforms")
plt.xlabel("Social Media Platform")
plt.ylabel("Total Engagement")
plt.show()
```

```python
# STEP 53: SAVE STATISTICAL RESULTS

statistical_results = pd.DataFrame({
    "Test": [test_name],
    "Statistic": [test_result.statistic],
    "P_Value": [test_result.pvalue],
    "Significance_Level": [alpha],
    "Decision": [
        "Reject H0" if test_result.pvalue < alpha
        else "Fail to reject H0"
    ]
})

display(statistical_results)

statistical_results.to_csv(
    "Statistical_Analysis_Results.csv",
    index=False
)

confidence_report.to_csv(
    "Confidence_Interval_Report.csv",
    index=False
)

print("Statistical reports saved successfully!")
```

```python
# STEP 55: Prepare the predictive modelling dataset

import pandas as pd
import numpy as np

# Work from the cleaned dataset
model_df = df.copy()

# Create the prediction target
model_df["total_engagement"] = (
    model_df["likes"] +
    model_df["comments"] +
    model_df["shares"]
)

# Extract features from the publication timestamp
model_df["post_hour"] = model_df["post_timestamp"].dt.hour
model_df["post_day"] = model_df["post_timestamp"].dt.day
model_df["post_month"] = model_df["post_timestamp"].dt.month
model_df["post_weekday"] = model_df["post_timestamp"].dt.dayofweek

# Select only potential pre-publication predictors
feature_columns = [
    "platform",
    "post_type",
    "post_hour",
    "post_day",
    "post_month",
    "post_weekday",
    "audience_age",
    "audience_gender",
    "audience_location",
    "audience_interests"
]

target_column = "total_engagement"

# Create modelling dataset
model_data = model_df[feature_columns + [target_column, "post_timestamp"]].copy()

# Remove rows where the target or timestamp is invalid
model_data = model_data.dropna(
    subset=[target_column, "post_timestamp"]
)

# Check for invalid engagement values
model_data = model_data[
    model_data[target_column] >= 0
]

# Display the results
print("Modelling dataset shape:", model_data.shape)

print("\nSelected predictor columns:")
print(feature_columns)

print("\nTarget variable:", target_column)

print("\nTarget variable statistics:")
display(model_data[target_column].describe())

print("\nMissing values in selected predictors:")
display(model_data[feature_columns].isnull().sum())

print("\nSample modelling records:")
display(model_data.head())

print("\nTarget distribution skewness:",
      round(model_data[target_column].skew(), 3))
```

```python
# STEP 56: Chronological train-test split

from sklearn.model_selection import train_test_split

# Sort records by publication timestamp
model_data = model_data.sort_values(
    by="post_timestamp"
).reset_index(drop=True)

# Calculate split position
split_index = int(len(model_data) * 0.80)

# Split into training and testing datasets
train_data = model_data.iloc[:split_index].copy()
test_data = model_data.iloc[split_index:].copy()

# Separate predictors and target
X_train = train_data[feature_columns]
X_test = test_data[feature_columns]

y_train = train_data[target_column]
y_test = test_data[target_column]

# Display split information
print("Training records:", len(X_train))
print("Testing records:", len(X_test))

print("\nTraining date range:")
print(train_data["post_timestamp"].min())
print("to")
print(train_data["post_timestamp"].max())

print("\nTesting date range:")
print(test_data["post_timestamp"].min())
print("to")
print(test_data["post_timestamp"].max())

print("\nTraining target mean:", round(y_train.mean(), 2))
print("Testing target mean:", round(y_test.mean(), 2))

print("\nTraining and testing shapes:")
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
```

```python
# STEP 57: Build a baseline regression model

from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Create baseline model
baseline_model = DummyRegressor(strategy="mean")

# Train the baseline model
baseline_model.fit(X_train, y_train)

# Generate predictions
baseline_predictions = baseline_model.predict(X_test)

# Evaluate baseline performance
baseline_mae = mean_absolute_error(
    y_test, baseline_predictions
)

baseline_rmse = np.sqrt(
    mean_squared_error(y_test, baseline_predictions)
)

baseline_r2 = r2_score(
    y_test, baseline_predictions
)

# Display results
print("BASELINE MODEL PERFORMANCE")
print("---------------------------")
print("MAE:", round(baseline_mae, 2))
print("RMSE:", round(baseline_rmse, 2))
print("R² Score:", round(baseline_r2, 4))
```

```python
# STEP 58: Build an improved Random Forest regression model

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestRegressor

# Identify numerical and categorical predictors
numeric_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("Numerical features:", numeric_features)
print("Categorical features:", categorical_features)

# Preprocessing for numerical columns
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median"))
])

# Preprocessing for categorical columns
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(
        handle_unknown="ignore",
        min_frequency=10
    ))
])

# Combine preprocessing steps
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

# Create the complete model pipeline
random_forest_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("regressor", RandomForestRegressor(
        n_estimators=150,
        max_depth=15,
        min_samples_split=10,
        min_samples_leaf=4,
        random_state=42,
        n_jobs=-1
    ))
])

# Train the model
random_forest_model.fit(X_train, y_train)

print("\nRandom Forest model trained successfully.")
```

```python
# STEP 59: Evaluate the improved model

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

# Generate predictions
rf_predictions = random_forest_model.predict(X_test)

# Calculate evaluation metrics
rf_mae = mean_absolute_error(y_test, rf_predictions)

rf_rmse = np.sqrt(
    mean_squared_error(y_test, rf_predictions)
)

rf_r2 = r2_score(y_test, rf_predictions)

# Create model comparison table
model_comparison = pd.DataFrame({
    "Model": ["Baseline Model", "Random Forest"],
    "MAE": [baseline_mae, rf_mae],
    "RMSE": [baseline_rmse, rf_rmse],
    "R2 Score": [baseline_r2, rf_r2]
})

print("MODEL PERFORMANCE COMPARISON")
display(model_comparison.round(4))

# Calculate percentage improvement in MAE
if baseline_mae != 0:
    mae_improvement = (
        (baseline_mae - rf_mae) / baseline_mae
    ) * 100

    print(
        "\nMAE improvement over baseline:",
        round(mae_improvement, 2),
        "%"
    )

# Interpret model performance
if rf_mae < baseline_mae:
    print("\nRandom Forest has lower MAE than the baseline.")
else:
    print("\nRandom Forest did not outperform the baseline on MAE.")

if rf_r2 > baseline_r2:
    print("Random Forest has a higher R² score than the baseline.")
else:
    print("Random Forest did not improve the R² score.")
```

```python
# STEP 66: Monthly forecasting baseline

from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

monthly_data = monthly_data.sort_values("Month").reset_index(drop=True)

if len(monthly_data) < 12:
    print("Not enough monthly observations for a meaningful forecast.")
else:
    split_point = int(len(monthly_data) * 0.80)

    train_monthly = monthly_data.iloc[:split_point].copy()
    test_monthly = monthly_data.iloc[split_point:].copy()

    # Naive forecast: use the latest training observation
    naive_predictions = np.repeat(
        train_monthly["Total_Engagement"].iloc[-1],
        len(test_monthly)
    )

    forecast_results = pd.DataFrame({
        "Month": test_monthly["Month"].values,
        "Actual": test_monthly["Total_Engagement"].values,
        "Naive_Prediction": naive_predictions
    })

    naive_mae = mean_absolute_error(
        forecast_results["Actual"],
        forecast_results["Naive_Prediction"]
    )

    naive_rmse = np.sqrt(mean_squared_error(
        forecast_results["Actual"],
        forecast_results["Naive_Prediction"]
    ))

    print("NAIVE FORECAST PERFORMANCE")
    print("MAE:", round(naive_mae, 2))
    print("RMSE:", round(naive_rmse, 2))

    # Seasonal naive forecast, using the value 12 months earlier
    if len(train_monthly) >= 12:
        seasonal_predictions = []

        history = train_monthly["Total_Engagement"].tolist()

        for i in range(len(test_monthly)):
            seasonal_predictions.append(history[-12])
            history.append(test_monthly["Total_Engagement"].iloc[i])

        forecast_results["Seasonal_Naive_Prediction"] = seasonal_predictions

        seasonal_mae = mean_absolute_error(
            forecast_results["Actual"],
            forecast_results["Seasonal_Naive_Prediction"]
        )

        seasonal_rmse = np.sqrt(mean_squared_error(
            forecast_results["Actual"],
            forecast_results["Seasonal_Naive_Prediction"]
        ))

        print("\nSEASONAL NAIVE FORECAST PERFORMANCE")
        print("MAE:", round(seasonal_mae, 2))
        print("RMSE:", round(seasonal_rmse, 2))

        forecast_comparison = pd.DataFrame({
            "Forecast Model": ["Naive", "Seasonal Naive"],
            "MAE": [naive_mae, seasonal_mae],
            "RMSE": [naive_rmse, seasonal_rmse]
        })

        display(forecast_comparison.round(2))
    else:
        print("\nNot enough training months for seasonal naive forecasting.")

    display(forecast_results.head(10))
```

```python
# STEP 72: Posting time analysis

time_performance = (
    model_df.groupby("post_hour")
    .agg(
        Number_of_Posts=("post_id", "count"),
        Average_Engagement=("total_engagement", "mean")
    )
    .reset_index()
)

weekday_performance = (
    model_df.groupby("post_weekday")
    .agg(
        Number_of_Posts=("post_id", "count"),
        Average_Engagement=("total_engagement", "mean")
    )
    .reset_index()
)

weekday_names = {
    0: "Monday",
    1: "Tuesday",
    2: "Wednesday",
    3: "Thursday",
    4: "Friday",
    5: "Saturday",
    6: "Sunday"
}

weekday_performance["Weekday"] = (
    weekday_performance["post_weekday"].map(weekday_names)
)

print("ENGAGEMENT BY POSTING HOUR")
display(time_performance.round(2))

print("\nENGAGEMENT BY WEEKDAY")
display(weekday_performance.round(2))

# Hour chart
plt.figure(figsize=(11, 5))
sns.lineplot(
    data=time_performance,
    x="post_hour",
    y="Average_Engagement",
    marker="o"
)
plt.title("Average Engagement by Posting Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Average Engagement")
plt.xticks(range(0, 24))
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# Weekday chart
plt.figure(figsize=(10, 5))
sns.barplot(
    data=weekday_performance,
    x="Weekday",
    y="Average_Engagement"
)
plt.title("Average Engagement by Weekday")
plt.xlabel("Weekday")
plt.ylabel("Average Engagement")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# STEP 73: Audience segment analysis

# Create age groups
model_df["audience_age_group"] = pd.cut(
    model_df["audience_age"],
    bins=[0, 18, 25, 35, 45, 55, 100],
    labels=[
        "Under 18",
        "18-25",
        "26-35",
        "36-45",
        "46-55",
        "Above 55"
    ],
    include_lowest=True
)

# Engagement by age group
age_performance = (
    model_df.groupby("audience_age_group", observed=False)
    .agg(
        Number_of_Posts=("post_id", "count"),
        Average_Engagement=("total_engagement", "mean"),
        Median_Engagement=("total_engagement", "median")
    )
    .reset_index()
)

# Engagement by gender
gender_performance = (
    model_df.groupby("audience_gender")
    .agg(
        Number_of_Posts=("post_id", "count"),
        Average_Engagement=("total_engagement", "mean"),
        Median_Engagement=("total_engagement", "median")
    )
    .reset_index()
)

print("ENGAGEMENT BY AGE GROUP")
display(age_performance.round(2))

print("\nENGAGEMENT BY AUDIENCE GENDER")
display(gender_performance.round(2))

# Age chart
plt.figure(figsize=(10, 5))
sns.barplot(
    data=age_performance,
    x="audience_age_group",
    y="Average_Engagement"
)
plt.title("Average Engagement by Audience Age Group")
plt.xlabel("Age Group")
plt.ylabel("Average Engagement")
plt.tight_layout()
plt.show()

# Gender chart
plt.figure(figsize=(8, 5))
sns.barplot(
    data=gender_performance,
    x="audience_gender",
    y="Average_Engagement"
)
plt.title("Average Engagement by Audience Gender")
plt.xlabel("Audience Gender")
plt.ylabel("Average Engagement")
plt.tight_layout()
plt.show()
```

```python
!pip -q install python-pptx

from pptx import Presentation
from pptx.util import Inches, Pt
from pptx.dml.color import RGBColor
from pptx.enum.text import PP_ALIGN
from pptx.enum.shapes import MSO_SHAPE
from pptx.enum.dml import MSO_THEME_COLOR

prs = Presentation()
prs.slide_width = Inches(13.333)
prs.slide_height = Inches(7.5)

# Theme
NAVY = RGBColor(25, 45, 75)
BLUE = RGBColor(45, 105, 165)
LIGHT = RGBColor(240, 244, 249)
DARK = RGBColor(40, 45, 55)
WHITE = RGBColor(255, 255, 255)
GREY = RGBColor(105, 115, 125)

def add_background(slide):
    shape = slide.shapes.add_shape(
        MSO_SHAPE.RECTANGLE, 0, 0,
        prs.slide_width, prs.slide_height
    )
    shape.fill.solid()
    shape.fill.fore_color.rgb = LIGHT
    shape.line.fill.background()
    slide.shapes._spTree.remove(shape._element)
    slide.shapes._spTree.insert(2, shape._element)

def add_text(slide, text, x, y, w, h,
             size=20, color=DARK, bold=False):
    box = slide.shapes.add_textbox(
        Inches(x), Inches(y), Inches(w), Inches(h)
    )
    tf = box.text_frame
    tf.word_wrap = True
    tf.margin_left = Inches(0.08)
    tf.margin_right = Inches(0.08)
    tf.margin_top = Inches(0.05)

    for i, line in enumerate(text.split("\n")):
        p = tf.paragraphs[0] if i == 0 else tf.add_paragraph()
        p.text = line
        p.font.size = Pt(size)
        p.font.bold = bold
        p.font.color.rgb = color
        p.space_after = Pt(12)
    return box

def add_slide(title, bullets, subtitle=None):
    slide = prs.slides.add_slide(prs.slide_layouts[6])
    add_background(slide)

    # Header
    header = slide.shapes.add_shape(
        MSO_SHAPE.RECTANGLE, 0, 0,
        prs.slide_width, Inches(1.05)
    )
    header.fill.solid()
    header.fill.fore_color.rgb = NAVY
    header.line.fill.background()

    add_text(slide, title, 0.55, 0.18, 12.2, 0.65,
             size=27, color=WHITE, bold=True)

    if subtitle:
        add_text(slide, subtitle, 0.7, 1.25, 11.9, 0.45,
                 size=15, color=GREY)

    add_text(slide, "\n".join("• " + b for b in bullets),
             0.8, 1.85, 11.8, 4.9, size=21)

    # Footer
    add_text(slide, "Social Media Engagement & Content Performance Analytics",
             0.55, 7.12, 11.8, 0.2, size=9, color=GREY)
    return slide

# Slide 1: Title
slide = prs.slides.add_slide(prs.slide_layouts[6])
add_background(slide)
add_text(slide, "SOCIAL MEDIA ENGAGEMENT",
         0.9, 1.7, 11.5, 0.8, size=34, color=NAVY, bold=True)
add_text(slide, "& CONTENT PERFORMANCE ANALYTICS",
         0.9, 2.55, 11.5, 0.7, size=27, color=BLUE, bold=True)
add_text(slide,
         "Integrated Decision Analysis and Responsible Reporting\n\n"
         "Data Analytics Capstone Project",
         0.95, 3.65, 10.8, 1.3, size=21, color=DARK)

# Slide 2
add_slide("Business Problem",
    [
        "Social media teams need to understand which platforms and content formats receive engagement.",
        "Large post-level datasets can reveal patterns in likes, comments, shares and engagement rates.",
        "The project examines performance patterns and evaluates whether engagement can be predicted.",
        "The analysis is descriptive and predictive, not proof of causal effects."
    ])

# Slide 3
add_slide("Project Objectives",
    [
        "Audit and prepare the social media dataset.",
        "Create engagement KPIs and time-based features.",
        "Explore engagement across platforms, content types and posting times.",
        "Compare groups using an appropriate statistical method.",
        "Evaluate a baseline and a machine-learning model.",
        "Explore monthly engagement forecasting and develop practical recommendations."
    ])

# Slide 4
add_slide("Dataset Overview",
    [
        "Total records: 19,885",
        "Total columns: 18",
        "Platforms represented: 4",
        "Post types represented: 3",
        "Dataset includes post timestamps, engagement metrics and audience attributes.",
        "Dataset provenance and representativeness require independent verification."
    ])

# Slide 5
add_slide("Data Preparation & KPI Engineering",
    [
        "Inspected data types, missing values, duplicates and numerical ranges.",
        "Standardised column names and text fields and parsed timestamps.",
        "Created total engagement from likes, comments and shares.",
        "Derived time features such as posting hour, weekday and month.",
        "Reviewed missing campaign and influencer identifiers and the empty sentiment field.",
        "Documented data-quality and interpretation limitations."
    ])

# Slide 6
add_slide("Engagement Performance",
    [
        "Total engagement: 16,886,205",
        "Average engagement per post: 849.19",
        "Average engagement rate: 25.48 (confirm the dataset's unit before using a percent sign).",
        "Twitter: 4,935 posts and average engagement of 856.09.",
        "Video: 6,613 posts and average engagement of 858.12.",
        "Insert your engagement distribution and platform/content comparison charts."
    ])

# Slide 7
add_slide("Platform & Content Findings",
    [
        "Twitter had the highest average engagement in the supplied platform summary.",
        "Video had the highest average engagement in the supplied content-type summary.",
        "Twitter–Video combination: 1,656 posts.",
        "Twitter–Video average engagement: 866.47; median engagement: 882.",
        "These differences are descriptive and do not demonstrate that platform or format caused engagement.",
        "Insert your platform/content heatmap and comparison charts."
    ])

# Slide 8
add_slide("Statistical Analysis",
    [
        "A group comparison was conducted as part of the analysis.",
        "Insert the two platform names, selected engagement metric, test name and test statistic.",
        "Insert the actual p-value and confidence intervals from your notebook.",
        "Explain the result using the chosen test's interpretation.",
        "Discuss assumptions, possible dependence between posts and practical significance.",
        "Do not claim statistical significance unless supported by the actual output."
    ])

# Slide 9
add_slide("Predictive Modelling",
    [
        "Baseline MAE: 271.12",
        "Random Forest MAE: 271.15",
        "Random Forest RMSE: 324.05",
        "Random Forest R²: −0.0007",
        "The Random Forest did not outperform the baseline.",
        "The current model should not be recommended for operational prediction."
    ])

# Slide 10
add_slide("Monthly Forecasting",
    [
        "Monthly observations: 37",
        "Held-out forecast test observations: 8",
        "Monthly engagement was aggregated into a time series.",
        "Naive and seasonal-naive forecasts were explored.",
        "Insert actual test-period forecast error metrics and the forecast chart.",
        "The short time series limits confidence in seasonality and long-term projections."
    ])

# Slide 11
add_slide("Recommendations & Responsible Reporting",
    [
        "Use platform and content summaries as exploratory guidance.",
        "Test video strategies across platforms rather than assuming one format always wins.",
        "Do not deploy the current Random Forest model without meaningful improvement.",
        "Validate forecast accuracy against simple benchmarks before planning with it.",
        "Protect audience privacy and avoid exposing identifiable or sensitive information.",
        "Distinguish association from causation and disclose dataset limitations."
    ])

# Slide 12
add_slide("Conclusion & Future Work",
    [
        "The project developed a workflow for data auditing, KPI engineering and engagement analysis.",
        "Twitter, video and the Twitter–Video combination showed strong average engagement in the supplied summaries.",
        "The Random Forest model did not improve on the baseline.",
        "Forecast evaluation needs actual test-period error metrics before conclusions can be made.",
        "Future work: verify data provenance, improve predictor quality, test more robust models and obtain a longer time series."
    ])

output_file = "Social_Media_Capstone_Final_Presentation.pptx"
prs.save(output_file)

print("Presentation created:", output_file)

from google.colab import files
files.download(output_file)
```
